```python
# DATALOADER TEST EXECUTION

try:
    # Point to flowers dataset folders (train/val/test)
    TEST_PARAMS = {
        'train_dir': os.path.join(PROJECT_ROOT, 'datasets', 'flowers', 'train'),
        'val_dir': os.path.join(PROJECT_ROOT, 'datasets', 'flowers', 'val'),
        'test_dir': os.path.join(PROJECT_ROOT, 'datasets', 'flowers', 'test'),
        'batch_size': 32 if torch.cuda.is_available() else 8,
        'augmentation': 'randaugment'  # Options: 'none', 'light', 'randaugment'
    }

    print("\n⚙️ Test Configuration:")
    for k, v in TEST_PARAMS.items():
        print(f"{k:>12}: {v}")

    # Load dataloaders — now returns 4 values (train, val, test, class_names)
    train_loader, val_loader, test_loader, class_names = get_dataloaders(**TEST_PARAMS)

    # Output basic stats
    print("\n✓ Dataloader Validation Passed:")
    print(f"- Class count: {len(class_names)}")
    print(f"- Classes: {class_names}")
    print(f"- Train batches: {len(train_loader)}")
    print(f"- Val batches: {len(val_loader)}")
    print(f"- Test batches: {len(test_loader)}")
    print(f"- Sample shape: {next(iter(train_loader))[0].shape}")

    import matplotlib.pyplot as plt
    import torchvision

    # Display a batch of augmented images
    def show_augmented_batch(loader, class_names):
        images, labels = next(iter(loader))
        grid = torchvision.utils.make_grid(images[:8], nrow=4, normalize=True, padding=2)
        plt.figure(figsize=(12, 6))
        plt.imshow(grid.permute(1, 2, 0).cpu().numpy())
        plt.title("Augmented Samples (Train Loader)")
        plt.axis("off")
        plt.show()

    # Show images from train_loader
    show_augmented_batch(train_loader, class_names)

    print(f"- Device: {'GPU' if torch.cuda.is_available() else 'CPU'} ready")

except Exception as e:
    print(f"\n✗ Load Failed: {e}")
    print("🔍 Debug Tips:")
    print(f"- Train dir exists: {os.path.exists(TEST_PARAMS['train_dir'])}")
    print(f"- Val dir exists: {os.path.exists(TEST_PARAMS['val_dir'])}")
    print(f"- Test dir exists: {os.path.exists(TEST_PARAMS['test_dir'])}")
    if os.path.exists(TEST_PARAMS['train_dir']):
        print(f"- Train classes: {os.listdir(TEST_PARAMS['train_dir'])}")
    if os.path.exists(TEST_PARAMS['val_dir']):
        print(f"- Val classes: {os.listdir(TEST_PARAMS['val_dir'])}")
    if os.path.exists(TEST_PARAMS['test_dir']):
        print(f"- Test classes: {os.listdir(TEST_PARAMS['test_dir'])}")
    raise
```
