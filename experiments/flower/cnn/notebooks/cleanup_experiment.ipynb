```python
# ⚠️ WARNING: This will delete all model artifacts for the current experiment


import os
import sys
import shutil
import importlib

# Setup correct root path
# PROJECT_ROOT = r'C:\Users\aliyu\Desktop\architectural-inductive-bias-study'
PROJECT_ROOT = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))

if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)
    

# Import config
CONFIG_MODULE = 'configs.flower_mlp'  # ← change if needed
config_module = importlib.import_module(CONFIG_MODULE)
CONFIG = config_module.CONFIG

# Experiment path
experiment_root = os.path.join(CONFIG['project_root'], 'experiments', CONFIG['dataset'], CONFIG['model_name'])

# Folders to remove
folders_to_remove = ['checkpoints', 'models', 'plots', 'tf-logs']
for folder in folders_to_remove:
    path = os.path.join(experiment_root, folder)
    if os.path.exists(path):
        shutil.rmtree(path)
        print(f"✓ Deleted: {path}")
    else:
        print(f"⚠️ Skipped (not found): {path}")
```
